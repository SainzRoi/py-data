```python
import json
with open(path) as f:
    records = [json.loads(line) for line in f]
records
```

```python
time_zones = [rec["tz"] for rec in records if "tz" in rec]
time_zones
```

## Pure Python Method

```python
#a function counting the number of timezones (method 1)(purepython)

def get_counts(sequence):
    counts = {}
    for x in sequence:
        if x in counts:
            counts[x] += 1
        else:
            counts[x] = 1
    return counts
```

```python
counts = get_counts(time_zones) # a dictionary containing (country, tz)(Method 1)
print(counts["America/New_York"])
print(len(time_zones))
```

```python
def top_counts(count_dict, n=10):   #function for finding the top 10 on the list
    value_key_pairs = [(count, tz) for tz, count in count_dict.items()] #changing the structure from the (time_zone, count) format to (count, time_zone) format so it can be easily sort 
    value_key_pairs.sort() #now it is now easier to sort. by default, it sorts into an ascending order
    return value_key_pairs[-n:] #since we want to find the top 10 and the current sort is in ascending order we want to sort it to descending order. 
```

```python
top_counts(counts)
```

```python
def top_counts(count_dict, n=10):   #alternatively with a cleaner descending format (roi's version)
    value_key_pairs = [(count, tz) for tz, count in count_dict.items()] #changing the structure from the (time_zone, count) format to (count, time_zone) format so it can be easily sort 
    value_key_pairs.sort(reverse=True) #now it is now easier to sort. now we provide the argument reverse=True so it would return a descending format.
    return value_key_pairs[:n]

import pandas as pd 

top = top_counts(counts, 10)

top_10 = (
    pd.Series(counts)
    .sort_values(ascending=False)
    .head(10)
)

top_10
```

```python
top_10 = top_10.rename_axis("Timezone")
top_10
```

```python
top_10 = top_10.reset_index(name="Count")
top_10
```

## defaultdict method

```python
#a function counting the number of timezones (method 2)(defaultdict)

from collections import defaultdict

def get_counts2(sequence):
    counts = defaultdict(int) #values will initialize to 0
    for x in sequence:
        counts[x] += 1
    return counts
```

```python
counts2 = get_counts(time_zones) # a dictionary containing (country, tz)(Method 2)
print(counts2["America/New_York"])
print(len(time_zones))
```
